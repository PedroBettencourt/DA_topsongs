### Drop irrelevant columns

```python
df = df.drop(['Artist URI(s)', 'Album URI', 'Album Artist URI(s)', 'Album Image URL', 'Track Preview URL', 'ISRC', 'Album Genres', 'Copyrights', 'Added By', 'Added At'], axis=1)
```

```python
df.info()
```

```python
# Drop 2 tracks without a name and artist

df.dropna(subset=['Track Name'], inplace=True)
df.reset_index(drop=True, inplace=True)
```

### Check null values outside of artist genres

```python
df[df.drop(['Artist Genres'], axis=1).isnull().any(axis=1)]
```

```python
# Considering it's only 8 songs out of 10000, I've decided to drop the songs that don't have information about danceability and similar parameters

df.dropna(subset=['Danceability'], inplace=True)

df[df.drop(['Artist Genres'], axis=1).isnull().any(axis=1)]
# The 3 remaining songs only have no label information
```

### Drop duplicates

```python
df.shape
```

```python
df[df.duplicated(keep=False)].sort_values('Track Name').head()
```

```python
df.drop_duplicates(keep='first', inplace=True)
df.reset_index(drop=True, inplace=True)

df.shape
# Removed 48 duplicates
```

# Analysis

## Artists with the most hits

```python
df.groupby(['Artist Name(s)']).size().sort_values(ascending=False)
```

```python
# Some artists are grouped together with a comma

df_artists = df.copy()

df_artists['Artist Name(s)'] = df_artists['Artist Name(s)'].str.split(',')
df_artists = df_artists.explode('Artist Name(s)').reset_index(drop=True)
```

```python
artist_hit_count = df_artists.groupby('Artist Name(s)').size().sort_values(ascending=False).head(20)
```

```python
# Plot size
fig, ax = plt.subplots(figsize=(13, 6))

# Draw plot
artist_hit_count.plot.bar(color='firebrick', alpha=0.8)
ax.set_title('Artists by Hits', fontdict={'size':18})
ax.set(ylabel='Number of Songs', xlabel='Artists')
plt.xticks(rotation=45, horizontalalignment='right', fontsize=10)

plt.plot()
```

## Genres

```python
df_genres = df.copy()
df_genres.dropna(subset=['Artist Genres'], inplace=True)
df_genres['Artist Genres'] = df_genres['Artist Genres'].str.split(',')
df_genres = df_genres.explode('Artist Genres').reset_index(drop=True)
```

```python
genre_count = df_genres.groupby('Artist Genres').size().sort_values(ascending=False)
```

```python
# Plot size
fig, ax = plt.subplots(figsize=(13, 6))

# Draw plot
genre_count.head(20).plot.bar(color='firebrick', alpha=0.8)
ax.set_title('Top Genres', fontdict={'size':18})
ax.set(ylabel='Number of Songs', xlabel='Genres')
plt.xticks(rotation=45, horizontalalignment='right', fontsize=10)

plt.plot()
```

## Popularity vs Danceability

```python
# Convert release date to datetime
df['Album Release Date'] =pd.to_datetime(df['Album Release Date'])

# Plot size
fig, ax = plt.subplots(figsize=(10, 6))

# Scatter plot of popularity vs danceability of songs from at least 2010 and with non null popularity
df[(df['Album Release Date'] > '2010') 
   & (df['Popularity'] > 0)
   ].plot.scatter(
        x='Popularity', y='Danceability', c='firebrick', alpha=0.8, ax=ax)

ax.set_title('Popularity vs Danceability', fontdict={'size':18})
ax.set(ylabel='Danceability', xlabel='Popularity')
plt.xticks(rotation=45, horizontalalignment='right', fontsize=10)

plt.plot()
```

## Genres by Popularity

```python
# Mean stats for every genre
genre_stats = df_genres.groupby('Artist Genres').mean()
genre_stats.reset_index(inplace=True)

# Popularity only
genre_popularity = genre_stats[['Artist Genres', 'Popularity']].sort_values(by='Popularity', ascending=False)
```

```python
# Genres with high popularity can have just 1 song, so the genres with at least 100 songs are the only ones looked at
genre_count_more = genre_count[genre_count > 100]
genre_popularity_more = genre_popularity[genre_popularity['Artist Genres'].isin(genre_count_more.index)]
```

```python
# Plot size
fig, ax = plt.subplots(figsize=(13, 6))

# Draw plot
genre_popularity_more.head(20).plot.bar(x='Artist Genres', y='Popularity', color='firebrick', alpha=0.8, ax=ax)
ax.set_title('Genres by Popularity', fontdict={'size':18})
ax.set(ylabel='Popularity', xlabel='Genres')
plt.xticks(rotation=45, horizontalalignment='right', fontsize=10)

plt.plot()
```
